## 🔎 Customer Data Filtering

The `customers` table was analyzed using SQL filtering conditions to identify customers located in New York City (NYC) with a credit limit greater than 100,000.

This demonstrates the use of the `WHERE` clause with multiple conditions.

```sql
USE classicmodels;

SELECT
    customerNumber,
    customerName,
    city,
    country,
    creditLimit
FROM customers
WHERE city = 'NYC'
  AND creditLimit > 100000;
```

### 💡 Interpretation

This query demonstrates how SQL filtering can be used to identify customers who meet specific business criteria. Combining location and credit-limit conditions allows analysts to focus on a targeted customer segment for further analysis.

## 👥 Employee Data Filtering

The `employees` table was analyzed to identify Sales Representatives assigned to office code 1.

This query demonstrates filtering based on both office assignment and job title.

```sql
USE classicmodels;

SELECT
    employeeNumber,
    firstName,
    lastName,
    jobTitle,
    officeCode
FROM employees
WHERE officeCode = 1
  AND jobTitle = 'Sales Rep';
```

### 💡 Interpretation

The query identifies employees who meet both specified criteria: working in office code 1 and having the job title of Sales Representative.

This type of filtering can support employee-level analysis and help organizations review sales teams by office location and role.

## 📊 SQL Aggregation Analysis

SQL aggregation functions were used to summarize business data and calculate key metrics.

The analysis includes counting records, calculating totals and averages, and identifying minimum and maximum values. These functions are commonly used to generate business KPIs and summarize large datasets.

```sql
USE classicmodels;

SELECT
    COUNT(*) AS total_customers,
    ROUND(AVG(creditLimit), 2) AS average_credit_limit,
    ROUND(MIN(creditLimit), 2) AS minimum_credit_limit,
    ROUND(MAX(creditLimit), 2) AS maximum_credit_limit
FROM customers;
```

### 💡 Interpretation

This aggregation query provides a high-level summary of customer credit limits. The calculated metrics can help analysts understand the overall customer credit profile and identify the range of credit limits across the customer base.

## 📊 Product Line Stock Analysis

The `products` table was analyzed to identify product lines containing more than 10 products with inventory levels above 5,000 units.

This analysis demonstrates the use of `GROUP BY`, `HAVING`, and aggregate functions to summarize product inventory at the product-line level.

```sql
USE classicmodels;

SELECT
    productLine,
    COUNT(*) AS product_count
FROM products
WHERE quantityInStock > 5000
GROUP BY productLine
HAVING product_count > 10;
```

### 💡 Interpretation

This analysis identifies product lines with a relatively large number of products having high inventory levels.

Such analysis can help businesses identify product categories with substantial stock availability and support inventory planning and monitoring.

## 💰 Product Line Pricing Analysis

The minimum buy price was calculated for each product line to compare pricing levels across product categories.

The results were filtered to include product lines with a minimum buy price greater than 5, then sorted by minimum price in descending order. `LIMIT` and `OFFSET` were also used to demonstrate controlled result selection.

```sql
USE classicmodels;

SELECT
    productLine,
    MIN(buyPrice) AS minimum_price
FROM products
GROUP BY productLine
HAVING minimum_price > 5
ORDER BY minimum_price DESC
LIMIT 5 OFFSET 3;
```

### 💡 Interpretation

This analysis compares the minimum product purchase price across product lines. Filtering and ranking the results helps focus on product categories with relatively higher minimum pricing levels.

The use of `LIMIT` and `OFFSET` demonstrates how SQL can be used to retrieve a specific portion of an ordered result set.

## 🔗 INNER JOIN Analysis

An `INNER JOIN` was used to combine the `productlines` and `products` tables through their common `productLine` field.

This allows product-level information to be analyzed together with its corresponding product category.

```sql
USE classicmodels;

SELECT
    pl.productLine,
    p.productName,
    p.buyPrice,
    p.MSRP
FROM productlines AS pl
INNER JOIN products AS p
    ON pl.productLine = p.productLine
ORDER BY pl.productLine, p.productName;
```

### 💡 Interpretation

The join combines product category information with individual product details.

This type of relational analysis is useful for comparing products within product lines and examining pricing information across categories.

## 🔗 RIGHT JOIN Analysis

A `RIGHT JOIN` was used to combine employee and office information based on the shared `officeCode`.

The analysis connects employee details with their corresponding office location, including city, state, and country.

```sql
USE classicmodels;

SELECT
    e.employeeNumber,
    e.firstName,
    e.lastName,
    e.jobTitle,
    e.officeCode,
    o.city,
    o.state,
    o.country
FROM offices AS o
RIGHT JOIN employees AS e
    ON o.officeCode = e.officeCode
ORDER BY e.officeCode, e.lastName;
```

### 💡 Interpretation

The join combines employee records with the office information associated with their office code.

This demonstrates how SQL joins can be used to bring together information stored across related tables and support organizational analysis by employee and location.

## 🧠 Common Table Expression (CTE)

A Common Table Expression (CTE) was used to calculate the average customer credit limit and identify customers whose credit limits are above the calculated average.

CTEs improve query organization by allowing a temporary named result set to be defined and referenced within the main query.

```sql
USE classicmodels;

WITH average_credit AS (
    SELECT AVG(creditLimit) AS avg_credit_limit
    FROM customers
)
SELECT
    customerName,
    creditLimit
FROM customers
WHERE creditLimit > (
    SELECT avg_credit_limit
    FROM average_credit
)
ORDER BY creditLimit DESC;
```

### 💡 Interpretation

The query identifies customers whose credit limits are higher than the overall average customer credit limit.

This type of analysis can help businesses identify customers with relatively higher credit capacity for customer segmentation and credit-related analysis.

## 🧠 CTE: Office & Employee Analysis

A Common Table Expression was used to create a temporary employee dataset before joining it with the `offices` table.

This approach demonstrates how CTEs can make multi-step SQL queries easier to organize and read.

```sql
USE classicmodels;

WITH office_employees AS (
    SELECT
        e.employeeNumber,
        e.firstName,
        e.lastName,
        e.jobTitle,
        e.officeCode
    FROM employees AS e
)
SELECT
    oe.employeeNumber,
    oe.firstName,
    oe.lastName,
    oe.jobTitle,
    oe.officeCode,
    o.city,
    o.country
FROM office_employees AS oe
INNER JOIN offices AS o
    ON oe.officeCode = o.officeCode
ORDER BY oe.officeCode, oe.lastName;
```

### 💡 Interpretation

The CTE creates a temporary employee dataset containing the required employee attributes. The resulting dataset is then joined with office information to associate each employee with their corresponding location.

This demonstrates how CTEs can improve query structure when working with multiple relational tables.

## 🔍 Subquery Analysis

A subquery was used to calculate the average customer credit limit and identify customers whose credit limits are above the overall average.

Subqueries allow one SQL query to use the result of another query as a condition or data source.

```sql
USE classicmodels;

SELECT
    customerNumber,
    customerName,
    country,
    creditLimit
FROM customers
WHERE creditLimit > (
    SELECT AVG(creditLimit)
    FROM customers
)
ORDER BY creditLimit DESC;
```

### 💡 Interpretation

The inner query calculates the average credit limit across all customers. The outer query then returns customers whose credit limits are above that average.

This demonstrates how subqueries can be used to perform comparisons against dynamically calculated values.

## 🔍 Subquery: Employees by Office Location

A subquery was used to identify employees working in offices located in **NYC or Paris**.

The inner query retrieves the office codes for the selected cities, and the outer query uses those office codes to filter employees.

```sql
USE classicmodels;

SELECT
    e.employeeNumber,
    e.firstName,
    e.lastName,
    e.jobTitle,
    e.officeCode
FROM employees AS e
WHERE e.officeCode IN (
    SELECT o.officeCode
    FROM offices AS o
    WHERE o.city IN ('NYC', 'Paris')
)
ORDER BY e.officeCode, e.lastName;
```

### 💡 Interpretation

The subquery first identifies the office codes associated with NYC and Paris. The outer query then retrieves employees assigned to those offices.

This demonstrates how subqueries can connect information across related tables without directly using a join in the main query.

## 🔍 Derived Table Analysis

A derived table was used to create a temporary result set containing information for a specific office.

The derived table was then queried to retrieve the office location.

```sql
USE classicmodels;

SELECT
    city
FROM (
    SELECT *
    FROM offices
    WHERE officeCode = 4
) AS office_data;
```

## 🧩 COALESCE & Nested Subquery

The `COALESCE()` function was used to replace missing HTML descriptions with a default text value.

A nested subquery was also used to identify the product lines associated with two selected products.

This demonstrates how SQL can handle missing values while filtering data through multiple levels of related information.

```sql
USE classicmodels;

SELECT
    textDescription,
    COALESCE(htmlDescription, 'missing value') AS html_description
FROM productlines
WHERE productLine IN (
    SELECT productLine
    FROM products
    WHERE productName IN (
        '1952 Alpine Renault 1300',
        '1996 Moto Guzzi 1100i'
    )
);
```

### 💡 Interpretation

The nested subquery identifies the product lines associated with the selected products. The outer query retrieves the corresponding product-line descriptions.

`COALESCE()` ensures that a missing HTML description is represented by a readable default value instead of a NULL value.

## 👁️ SQL View — Customer Payment Summary

A SQL view named `amount_view` was created to provide a summarized view of customer payment activity.

The view calculates the number of payments, total payment amount, and average payment amount for each customer.

Views are useful for simplifying repeated analysis and presenting frequently used business metrics in a structured format.

```sql
USE classicmodels;

CREATE OR REPLACE VIEW amount_view AS
SELECT
    customerNumber,
    COUNT(*) AS payment_count,
    SUM(amount) AS total_payment_amount,
    AVG(amount) AS average_payment_amount
FROM payments
GROUP BY customerNumber;
```

```sql
SELECT *
FROM amount_view
ORDER BY total_payment_amount DESC;
```

### 💡 Interpretation

The view provides a reusable customer-level summary of payment activity. Instead of repeatedly writing the aggregation query, analysts can query the view directly when they need payment summaries.

This demonstrates how SQL views can improve query reusability and simplify reporting workflows.

## ⚙️ Stored Procedure — Employees by Office

A stored procedure named `employees_by_office` was created to retrieve employee details based on an office code.

Stored procedures allow reusable SQL logic to be stored in the database and executed with different input values.

```sql
USE classicmodels;

DROP PROCEDURE IF EXISTS employees_by_office;

DELIMITER $$

CREATE PROCEDURE employees_by_office(IN p_officeCode VARCHAR(10))
BEGIN
    SELECT
        employeeNumber,
        firstName,
        lastName,
        jobTitle,
        officeCode
    FROM employees
    WHERE officeCode = p_officeCode
    ORDER BY lastName, firstName;
END $$

DELIMITER ;
```

```sql
CALL employees_by_office('1');
```

### 💡 Interpretation

The stored procedure accepts an office code as an input parameter and returns the employees assigned to that office.

This demonstrates how stored procedures can make frequently used database operations reusable and parameter-driven.

## 🧮 User-Defined Function — Product Discount

A user-defined SQL function named `product_discount` was created to calculate a discount amount based on a product's MSRP.

For this demonstration, the function applies a 10% discount rule to the supplied price.

This demonstrates how reusable business logic can be implemented through SQL functions.

```sql
USE classicmodels;

DROP FUNCTION IF EXISTS product_discount;

DELIMITER $$

CREATE FUNCTION product_discount(price DECIMAL(10,2))
RETURNS DECIMAL(10,2)
DETERMINISTIC
BEGIN
    RETURN price * 0.10;
END $$

DELIMITER ;
```

```sql
SELECT
    productLine,
    quantityInStock,
    MSRP,
    product_discount(MSRP) AS discount_amount
FROM products;
```

### 💡 Interpretation

The function calculates a standardized discount amount from the product MSRP.

This demonstrates how user-defined functions can encapsulate reusable calculations and apply consistent business rules across SQL queries.

The 10% discount is a demonstration rule created for this project and does not represent an actual ClassicModels pricing policy.

## 🧮 User-Defined Function — Payment Discount

A second user-defined SQL function named `amount_disc` was created to calculate a discount amount based on a payment value.

For this demonstration, the function applies a 5% calculation to the supplied payment amount.

This demonstrates how reusable calculations can be applied directly within SQL queries.

```sql
USE classicmodels;

DROP FUNCTION IF EXISTS amount_disc;

DELIMITER $$

CREATE FUNCTION amount_disc(payment_amount DECIMAL(10,2))
RETURNS DECIMAL(10,2)
DETERMINISTIC
BEGIN
    RETURN payment_amount * 0.05;
END $$

DELIMITER ;
```

```sql
SELECT
    customerNumber,
    amount,
    amount_disc(amount) AS discount_amount
FROM payments;
```

### 💡 Interpretation

The function applies a standardized calculation to payment amounts and returns the resulting discount value.

This demonstrates how user-defined functions can be used to encapsulate reusable calculations within SQL queries.

The 5% calculation is a demonstration rule created for this project and does not represent an actual ClassicModels business policy.

## 🔔 SQL Trigger — Salary Change Logging

A database trigger was implemented to automatically record salary changes whenever an employee's salary was updated.

The trigger captures the employee ID, previous salary, new salary, and the date of the change.

This demonstrates how database triggers can automatically execute actions in response to data modifications.

```sql
USE sampledb;

CREATE TABLE IF NOT EXISTS employees (
    id INT PRIMARY KEY,
    name VARCHAR(100),
    salary DECIMAL(10,2)
);

CREATE TABLE IF NOT EXISTS salary_log (
    employee_id INT,
    old_salary DECIMAL(10,2),
    new_salary DECIMAL(10,2),
    change_date DATETIME
);

DELIMITER $$

CREATE TRIGGER after_salary_update
AFTER UPDATE ON employees
FOR EACH ROW
BEGIN
    INSERT INTO salary_log (
        employee_id,
        old_salary,
        new_salary,
        change_date
    )
    VALUES (
        OLD.id,
        OLD.salary,
        NEW.salary,
        NOW()
    );
END $$

DELIMITER ;
```

```sql
UPDATE employees
SET salary = 80000
WHERE id = 1;

SELECT *
FROM salary_log;
```

### 💡 Interpretation

The trigger automatically records salary changes in the `salary_log` table whenever an employee record is updated.

This demonstrates how triggers can support automated audit logging and maintain a history of important database changes.

The trigger uses a separate `sampledb` demonstration database because the ClassicModels `employees` table does not contain a salary field.

## 💰 Top 10 Customers by Total Payments

Customer payment activity was analyzed by joining the `customers` and `payments` tables.

The total payment amount was calculated for each customer, allowing the highest-value customers based on recorded payments to be identified.

```sql
USE classicmodels;

SELECT
    c.customerNumber,
    c.customerName,
    c.country,
    SUM(p.amount) AS total_payments
FROM customers AS c
INNER JOIN payments AS p
    ON c.customerNumber = p.customerNumber
GROUP BY
    c.customerNumber,
    c.customerName,
    c.country
ORDER BY total_payments DESC
LIMIT 10;
```

### 💡 Business Interpretation

This analysis identifies the customers with the highest total recorded payments.

The results can also be used as a starting point for deeper customer-value analysis.

## 💰 Revenue Analysis by Product Line

Revenue was analyzed at the product-line level by joining the `products` and `orderdetails` tables.

Total revenue was calculated using the ordered quantity and selling price for each product.

This analysis helps identify product categories that contribute the most to overall sales revenue.

```sql
USE classicmodels;

SELECT
    p.productLine,
    SUM(od.quantityOrdered * od.priceEach) AS total_revenue
FROM products AS p
INNER JOIN orderdetails AS od
    ON p.productCode = od.productCode
GROUP BY p.productLine
ORDER BY total_revenue DESC;
```

### 💡 Business Interpretation

This analysis ranks product lines according to their total recorded sales revenue.

Identifying high-revenue product lines can help businesses understand which categories contribute most to sales and can support product portfolio and sales planning decisions.

## 🏆 Top 10 Products by Revenue

Product-level revenue was analyzed by joining the `products` and `orderdetails` tables.

Total revenue was calculated for each product based on the quantity ordered and selling price. The top 10 products were then identified by ranking them according to total revenue.

```sql
USE classicmodels;

SELECT
    p.productCode,
    p.productName,
    p.productLine,
    SUM(od.quantityOrdered * od.priceEach) AS total_revenue
FROM products AS p
INNER JOIN orderdetails AS od
    ON p.productCode = od.productCode
GROUP BY
    p.productCode,
    p.productName,
    p.productLine
ORDER BY total_revenue DESC
LIMIT 10;
```

### 💡 Business Interpretation

The analysis is based on recorded order revenue and does not account for product costs or profit margins.

## 👤 Sales Performance by Employee

Sales performance was analyzed at the Sales Representative level by connecting employees with their assigned customers, orders, and order details.

Total sales revenue was calculated for each Sales Representative and ranked from highest to lowest.

```sql
USE classicmodels;

SELECT
    e.employeeNumber,
    e.firstName,
    e.lastName,
    SUM(od.quantityOrdered * od.priceEach) AS total_sales
FROM employees AS e
INNER JOIN customers AS c
    ON e.employeeNumber = c.salesRepEmployeeNumber
INNER JOIN orders AS o
    ON c.customerNumber = o.customerNumber
INNER JOIN orderdetails AS od
    ON o.orderNumber = od.orderNumber
WHERE e.jobTitle = 'Sales Rep'
GROUP BY
    e.employeeNumber,
    e.firstName,
    e.lastName
ORDER BY total_sales DESC;
```

### 💡 Business Interpretation

This analysis compares sales revenue generated through individual Sales Representatives.

Ranking Sales Representatives by recorded sales can help businesses evaluate sales performance, identify high-performing representatives, and support sales-team analysis.

The calculation represents recorded order revenue and does not account for product costs, discounts, or profit margins.

## 📦 Order Status Analysis

Order status distribution was analyzed to understand the overall composition of orders in the database.

The analysis calculates the number of orders for each status and the percentage that each status represents of all recorded orders.

```sql
USE classicmodels;

SELECT
    status,
    COUNT(*) AS order_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM orders),
        2
    ) AS percentage_of_orders
FROM orders
GROUP BY status
ORDER BY order_count DESC;
```

### 💡 Business Interpretation

This analysis provides an overview of the distribution of orders across different statuses.

Understanding order-status patterns can help businesses monitor order processing, identify frequently occurring statuses, and support operational performance analysis.

## 💵 Customer Revenue Analysis

Customer-level revenue was analyzed by combining the `customers`, `orders`, and `orderdetails` tables.

The analysis calculates the total number of orders, total revenue, and average order value for each customer. The top 10 customers were ranked by total revenue.

```sql
USE classicmodels;

SELECT
    c.customerNumber,
    c.customerName,
    c.country,
    COUNT(DISTINCT o.orderNumber) AS total_orders,
    ROUND(SUM(od.quantityOrdered * od.priceEach), 2) AS total_revenue,
    ROUND(
        SUM(od.quantityOrdered * od.priceEach) /
        COUNT(DISTINCT o.orderNumber),
        2
    ) AS average_order_value
FROM customers AS c
INNER JOIN orders AS o
    ON c.customerNumber = o.customerNumber
INNER JOIN orderdetails AS od
    ON o.orderNumber = od.orderNumber
GROUP BY
    c.customerNumber,
    c.customerName,
    c.country
ORDER BY total_revenue DESC
LIMIT 10;
```

### 💡 Business Interpretation

This analysis identifies customers generating the highest recorded order revenue and provides additional context through order count and average order value.

These metrics can support customer segmentation, account prioritization, and analysis of customer purchasing behavior.

The revenue calculation is based on recorded order quantities and selling prices and does not represent profit.

## 🌍 Sales Performance by Country

Sales performance was analyzed at the country level by combining customer, order, and order-detail information.

The analysis calculates the number of customers, number of orders, total revenue, and average order value for each country.

```sql
USE classicmodels;

SELECT
    c.country,
    COUNT(DISTINCT c.customerNumber) AS customer_count,
    COUNT(DISTINCT o.orderNumber) AS order_count,
    ROUND(SUM(od.quantityOrdered * od.priceEach), 2) AS total_revenue,
    ROUND(
        SUM(od.quantityOrdered * od.priceEach) /
        COUNT(DISTINCT o.orderNumber),
        2
    ) AS average_order_value
FROM customers AS c
INNER JOIN orders AS o
    ON c.customerNumber = o.customerNumber
INNER JOIN orderdetails AS od
    ON o.orderNumber = od.orderNumber
GROUP BY c.country
ORDER BY total_revenue DESC;
```

### 💡 Business Interpretation

This analysis provides a country-level view of customer and sales performance.

Comparing customer volume, order volume, total revenue, and average order value can help identify markets with stronger commercial activity and support regional sales planning.

## 📦 Top Orders by Order Value

Individual order values were calculated by combining the `orders` and `orderdetails` tables.

The total value of each order was calculated using the ordered quantity multiplied by the selling price. The top 10 highest-value orders were then identified.

```sql
USE classicmodels;

SELECT
    o.orderNumber,
    o.orderDate,
    o.status,
    ROUND(
        SUM(od.quantityOrdered * od.priceEach),
        2
    ) AS order_value
FROM orders AS o
INNER JOIN orderdetails AS od
    ON o.orderNumber = od.orderNumber
GROUP BY
    o.orderNumber,
    o.orderDate,
    o.status
ORDER BY order_value DESC
LIMIT 10;
```

### 💡 Business Interpretation

This analysis identifies the highest-value individual orders based on recorded sales value.

Analyzing large orders can help businesses understand major transactions and support order-level sales monitoring.

The calculated order value represents sales revenue based on quantity ordered and selling price; it does not represent profit.
